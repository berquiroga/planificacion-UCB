## 1. Modularidad: Encapsulando la Física
Para no escribir la misma matriz 1000 veces, la empaquetamos en una función de Python que reciba la frecuencia `w` como parámetro de entrada.

```python
import numpy as np
import matplotlib.pyplot as plt

# TAREA 1: Defina la función que construye y resuelve la matriz para cualquier w
def resolver_red(w):
    # Parámetros fijos
    R = 50.0
    L = 0.1   # 100 mH
    C = 1e-4  # 100 uF
    
    # Reactancias dinámicas
    XL = w * L
    XC = 1 / (w * C)
    
    # Matriz Z (2x2) y Vector V
    Z = np.array([
        [R + 1j*XL, -1j*XL],
        [-1j*XL, 20 + 1j*(XL - XC)] # Malla 2 acoplada
    ])
    V = np.array([10+0j, 0+0j])
    
    # Resolver y devolver la corriente de la malla 2 (I[1])
    I = np.linalg.solve(Z, V)
    return np.abs(I[1])
```

## 2. El Espectro: Vectores Logarítmicos
Generaremos un vector de frecuencias desde $10^1$ (10 rad/s) hasta $10^4$ (10,000 rad/s) usando `np.logspace`.

```python
# TAREA 2: Generar vector w y evaluar la función en bucle
w_array = np.logspace(1, 4, 500) # 500 puntos logarítmicos

I_malla2_array = []

# Bucle para barrer todas las frecuencias
for w in w_array:
    corriente_mag = resolver_red(w)
    I_malla2_array.append(corriente_mag)
```

## 3. Visualización de Ingeniería (Semilogx)
Graficaremos la respuesta del sistema para encontrar visualmente la resonancia.

```python
# TAREA 3: Crear el Dashboard
plt.figure(figsize=(10, 6))
# Utilizamos semilogx para que el eje X (w) sea logarítmico
plt.semilogx(w_array, I_malla2_array, color='b', linewidth=2)

plt.title('Respuesta en Frecuencia del Actuador (Magnitud de Corriente)', fontsize=14)
plt.xlabel('Frecuencia Angular $\omega$ (rad/s) [Escala Logarítmica]', fontsize=12)
plt.ylabel('Magnitud de Corriente $|I_2|$ (Amperios)', fontsize=12)
plt.grid(True, which="both", ls="--")

# Encontrar el pico máximo computacionalmente
max_corriente = np.max(I_malla2_array)
indice_max = np.argmax(I_malla2_array)
w_resonancia = w_array[indice_max]

print(f"\n>>> DIAGNÓSTICO DIGITAL <<<")
print(f"Corriente máxima: {max_corriente:.4f} A")
print(f"Detectada a la frecuencia de resonancia: {w_resonancia:.2f} rad/s")

# Guardar gráfico en HD (Dashboarding)
plt.savefig('barrido_frecuencial.png', dpi=300)
plt.show()
```
