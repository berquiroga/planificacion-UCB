## 1. El Puente Físico-Matemático

En matemáticas, un sistema de ecuaciones es abstracto. En ingeniería mecatrónica, el sistema $\mathbf{R}\mathbf{I} = \mathbf{V}$ representa el **punto de equilibrio energético** de nuestro hardware. La matriz $\mathbf{R}$ es el mapa de interconexión (topología), el vector $\mathbf{V}$ representa las fuentes de energía y el vector $\mathbf{I}$ es nuestra incógnita principal: el flujo real de carga.

## 2. LVK como Conservación de Energía

Aplicamos la Ley de Voltajes de Kirchhoff ($\sum V = 0$). Piense en la analogía de un campo conservativo (un excursionista en la montaña): las baterías elevan el potencial del sistema (ganancia) y las resistencias lo disipan en forma de calor (caída). Al cerrar un lazo, el balance energético neto debe ser nulo.

Estructurando la matriz canónica $\mathbf{R}$:
- **Impedancia Propia ($R_{kk}$):** Suma total de resistencia en el lazo $k$ (Diagonal, estrictamente positiva).
- **Impedancia Mutua ($R_{kj}$):** Resistencia compartida entre fronteras (Fuera de la diagonal, negativa por convención horaria).

```python
import numpy as np

# Configuración de impresión
np.set_printoptions(precision=2, suppress=True)
```

## 3. Implementación en Python

**Tarea:** Traduzca el modelo analítico (desarrollado manuscritamente en la guía) al entorno computacional.

```python
# SOLUCIÓN
R1, R2, R3, R4, R5, R6 = 10, 22, 15, 33, 47, 12
R = np.array([
    [R1 + R2 + R3, -R2, -R3],
    [-R2, R2 + R4 + R5, -R5],
    [-R3, -R5, R3 + R5 + R6]
])
V_fuentes = np.array([24.0, 0.0, -12.0])
print("Matriz de Resistencias R (Ohms):\n", R)
print("\nVector de Voltajes V (Volts):\n", V_fuentes)
```

## 4. Validación Automática (CDIO: Verificar)

Verificaremos las firmas estructurales de una red pasiva: simetría y diagonal positiva.

```python
# Validación Automática
assert R.ndim == 2, "Error: La matriz R debe ser bidimensional"
assert R.shape == (3, 3), "Error: La matriz R debe ser de 3x3"

# Verificación de simetría (R_kj == R_jk)
assert np.array_equal(R, R.T), "CRÍTICO: La matriz no es simétrica. Revise las impedancias mutuas."

# Verificación de la diagonal principal (R_kk > 0)
diagonal = np.diag(R)
assert np.all(diagonal > 0), "CRÍTICO: La resistencia propia de la malla debe ser estrictamente positiva."

print("✅ Certificación exitosa: Matriz simétrica y topológicamente viable.")
```
