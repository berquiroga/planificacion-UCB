## 1. El Operador Complejo en Python

Para denotar la parte imaginaria de un número (la *reactancia* $X$), Python emplea el sufijo `j`.

```python
import numpy as np
Z1 = 50 + 25j 
print(f"Impedancia Z1: {Z1} Ohmios")
```

## 2. Conversión a Forma Polar Computarizada

Utilizaremos `np.abs()` y `np.angle()`.

```python
magnitud_Z1 = np.abs(Z1)
fase_Z1_rad = np.angle(Z1)
print(f"Magnitud: {magnitud_Z1:.2f} | Fase (deg): {np.degrees(fase_Z1_rad):.2f}")
```

## 3. División Fasorial (Ley de Ohm)

Python permite operaciones algebraicas directas con complejos.

```python
V = 100 + 0j
I = V / Z1
print(f"Corriente I: {I} A")
```

## 4. El Salto Industrial: Sistemas Matriciales en CA

El mismo solver que usamos para mallas resistivas en Corriente Continua (`np.linalg.solve`) procesa nativamente matrices de impedancia compleja $\mathbf{Z}$. Resolvamos el sistema de $2 \times 2$ planteado en el sílabo:

$$ \mathbf{Z} = \begin{bmatrix} 50 + j25 & -j25 \\ -j25 & 20 - j10 \end{bmatrix} \Omega \quad ; \quad \mathbf{V} = \begin{bmatrix} 10\angle 0^\circ \\ 0 \end{bmatrix} \text{V} $$

```python
Z_mat = np.array([
    [50 + 25j, -25j],
    [-25j, 20 - 10j]
])

V_vec = np.array([10 + 0j, 0 + 0j])

try:
    I_vec = np.linalg.solve(Z_mat, V_vec)
    print("Vector de Corrientes Fasoriales I:\n", I_vec)
    
    mag_I1 = np.abs(I_vec[0])
    fase_I1 = np.degrees(np.angle(I_vec[0]))
    print(f"\nI_1: {mag_I1:.4f} A con ángulo de {fase_I1:.2f}°")
except np.linalg.LinAlgError:
    print("Matriz Singular.")
```
