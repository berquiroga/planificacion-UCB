## 1. El Operador Complejo en Python

A diferencia de otros lenguajes de programación donde se requieren librerías de terceros para manejar matemáticas de ingeniería, Python fue adoptado ampliamente por la industria científica porque posee soporte nativo para el plano complejo.

Para denotar la parte imaginaria de un número (la *reactancia* $X$ en nuestro contexto), Python emplea el sufijo `j` (estándar de ingeniería eléctrica).

```python
import numpy as np

# SOLUCIÓN
Z1 = 50 + 25j 

print(f"Impedancia Z1 en forma rectangular: {Z1} Ohmios")
print(f"Tipo de dato: {type(Z1)}")
```

## 2. Conversión a Forma Polar Computarizada

Para validar nuestros cálculos a pulso (Pitágoras y ArcoTangente), utilizaremos dos funciones fundamentales de la librería NumPy:
- `np.abs()`: Extrae la magnitud $|\mathbf{Z}|$.
- `np.angle()`: Extrae la fase $\theta$ en radianes (podemos pedirla en grados).

```python
# SOLUCIÓN
magnitud_Z1 = np.abs(Z1) 
fase_Z1_rad = np.angle(Z1) 

fase_Z1_deg = np.degrees(fase_Z1_rad) 

print(f"Magnitud (|Z1|): {magnitud_Z1:.2f} Ohmios")
print(f"Fase (Theta): {fase_Z1_deg:.2f} Grados")
```

## 3. División Fasorial (Ley de Ohm)

Python permite realizar operaciones algebraicas directamente con las variables complejas, manejando la trigonometría en segundo plano por nosotros.

```python
# SOLUCIÓN
V = 100 + 0j

I = V / Z1

magnitud_I = np.abs(I)
fase_I = np.angle(I)

print(f"Corriente I (Rectangular): {I} Amperios")
print(f"Magnitud de Corriente (|I|): {magnitud_I:.2f} A")
print(f"Desfase de Corriente: {np.degrees(fase_I):.2f} Grados")
```
